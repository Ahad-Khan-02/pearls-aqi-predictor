```python
plt.figure(figsize=(15,6))

plt.plot(df["timestamp"], df["aqi"])

plt.title("AQI Over Time")
plt.xlabel("Timestamp")
plt.ylabel("AQI")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(14,10))

correlation_matrix = df.corr(numeric_only=True)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation Heatmap")

plt.show()

correlation_matrix["future_aqi"].sort_values(ascending=False)
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(x=df["hour"], y=df["aqi"])

plt.title("AQI by Hour of Day")

plt.xlabel("Hour")
plt.ylabel("AQI")

plt.show()
```

# Advanced Time-Series EDA

###### CONVERT TIMESTAMP PROPERLY

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

##### DAILY AQI TREND

```python
daily_aqi = df.groupby(df["timestamp"].dt.date)["aqi"].mean()

plt.figure(figsize=(15,6))

plt.plot(daily_aqi.index, daily_aqi.values)

plt.title("Daily Average AQI Trend")
plt.xlabel("Date")
plt.ylabel("Average AQI")

plt.xticks(rotation=45)

plt.show()
```

##### MONTHLY AQI ANALYSIS

```python
monthly_aqi = df.groupby("month")["aqi"].mean()

plt.figure(figsize=(10,5))

sns.barplot(
    x=monthly_aqi.index,
    y=monthly_aqi.values
)

plt.title("Average AQI by Month")
plt.xlabel("Month")
plt.ylabel("Average AQI")

plt.show()
```

##### WEEKDAY VS WEEKEND ANALYSIS

```python
weekday_names = {
    0: "Mon",
    1: "Tue",
    2: "Wed",
    3: "Thu",
    4: "Fri",
    5: "Sat",
    6: "Sun"
}

df["weekday_name"] = df["day_of_week"].map(weekday_names)

plt.figure(figsize=(10,5))

sns.boxplot(
    x="weekday_name",
    y="aqi",
    data=df
)

plt.title("AQI by Day of Week")

plt.show()
```

##### POLLUTANT IMPACT ANALYSIS

```python
pollutants = ["pm25", "pm10", "no2", "o3"]

for pollutant in pollutants:

    plt.figure(figsize=(8,5))

    sns.scatterplot(
        x=df[pollutant],
        y=df["aqi"]
    )

    plt.title(f"{pollutant} vs AQI")

    plt.xlabel(pollutant)
    plt.ylabel("AQI")

    plt.show()
```

##### MOVING AVERAGE ANALYSIS

```python
df["aqi_24hr_avg"] = df["aqi"].rolling(window=24).mean()

plt.figure(figsize=(15,6))

plt.plot(df["timestamp"], df["aqi"], label="Original AQI")

plt.plot(
    df["timestamp"],
    df["aqi_24hr_avg"],
    label="24-Hour Moving Average"
)

plt.title("AQI with Moving Average")

plt.xlabel("Timestamp")
plt.ylabel("AQI")

plt.legend()

plt.show()
```

##### OUTLIER DETECTION

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df["aqi"])

plt.title("AQI Outlier Detection")

plt.show()
```

##### AUTOCORRELATION ANALYSIS

```python
from statsmodels.graphics.tsaplots import plot_acf

plt.figure(figsize=(12,6))

plot_acf(df["aqi"], lags=50)

plt.title("AQI Autocorrelation")

plt.show()
```
